```python
import numpy as np
```

```python
## Transition matrix considering R, A and F as states
p = 0.3
q = 0.7
P = np.array([[1-p, p, 0],
            [0, 0, 1],
            [q, 0, 1-q]])
```

```python
print(P)
```

```python
## Neuron initial state
x_0 = np.array([1, 0, 0])
```

```python
## Helpers
def calc_matrix(P, x, n):
    for _ in range(n):
        x = x @ P
    return x

def vals_m(x):
    labels = ["R", "A", "F"]
    for i in range(len(x)):
        return "\n".join([f"{labels[i]}: {round(x[i], 2)}" for i in range(len(x))])
```

```python
x_1 = calc_matrix(P, x_0, 1)
x_2 = calc_matrix(P, x_0, 2)
x_3 = calc_matrix(P, x_0, 3)

print("Neuron state after 1 step:\n{}".format(vals_m(x_1)))
print("Neuron state after 2 steps:\n{}".format(vals_m(x_2)))
print("Neuron state after 3 steps:\n{}".format(vals_m(x_3)))
```

```python
## Stationary state
def find_stationary(P, x0, tol=1e-6, max_iter=1000):
    x = x0.copy()

    for i in range(max_iter):
        x_next = x @ P

        # measure difference
        if np.linalg.norm(x_next - x) < tol:
            print(f"Converged in {i} iterations")
            return x_next

        x = x_next

    print("Did not converge")
    return x
```

```python
x_stationary = find_stationary(P, x_0)
print(vals_m(x_stationary))
```
