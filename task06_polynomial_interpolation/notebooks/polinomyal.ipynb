```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# sample points we want the polynomial to pass through
x_pts = np.array([1, 2, 3])
y_pts = np.array([2, 3, 5])

# find the polynomial (degree 2) that fits these points
# this is basically solving the Vandermonde system behind the scenes
coeffs = np.polyfit(x_pts, y_pts, 2)

# turn the coefficients into an actual function
poly = np.poly1d(coeffs)
```

```python
# create a range to draw the curve
x = np.linspace(0, 4, 100)
y = poly(x)
```

```python
# plot the curve
plt.figure()
plt.plot(x, y, label="Interpolating polynomial")

# plot the original points
plt.scatter(x_pts, y_pts, color='red', zorder=5, label="Given points")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Polynomial interpolation (Vandermonde)")
plt.legend()
plt.grid(True)
plt.savefig("../figures/polynomial_interpolation.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
# print the polynomial
print("Interpolating polynomial:")
print(poly)

# check if it hits the points
print("\nChecking the points:")
for xi, yi in zip(x_pts, y_pts):
    print(f"x = {xi}, p(x) = {poly(xi):.2f}, expected = {yi}")
```
